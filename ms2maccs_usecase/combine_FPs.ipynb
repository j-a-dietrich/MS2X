```python
import pickle
import matplotlib.pyplot as plt
import numpy as np
from matchms.importing import load_from_mgf
from matchms.exporting import save_as_mgf
from tqdm import tqdm
from rdkit import Chem
from rdkit.Chem import MACCSkeys
```

## Fragment count vs annotation count

```python
with open("fp_bit_map_train_H_p_mode.pkl", 'rb') as f:
    fp_bit_map1 = pickle.load(f)

with open("fp_bit_map_val_H_p_mode.pkl", "rb") as f:
    fp_bit_map2 = pickle.load(f)

with open("fp_bit_map_test_H_p_mode.pkl", "rb") as f:
    fp_bit_map3 = pickle.load(f)
```

```python
fp_bit_map_comb = {}
all_keys = set(fp_bit_map1.keys()).union(set(fp_bit_map2.keys())).union(set(fp_bit_map3.keys()))

for key in all_keys:
    arrays = []
    if key in fp_bit_map1:
        arrays.append(fp_bit_map1[key])
    if key in fp_bit_map2:
        arrays.append(fp_bit_map2[key])
    if key in fp_bit_map3:
        arrays.append(fp_bit_map3[key])

    # Sum all arrays element-wise
    if arrays:
        fp_bit_map_comb[key] = np.sum(arrays, axis=0)
```

```python
len(fp_bit_map1.keys())
```

```python
len(all_keys)
```

```python
with open("fp_bit_map_H_p_mode.pkl", "wb") as f:
    pickle.dump(fp_bit_map_comb, f)
```
